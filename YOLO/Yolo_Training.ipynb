```python
# 1. 필수 라이브러리 설치
!pip install -q ultralytics roboflow

from roboflow import Roboflow
from ultralytics import YOLO

# 2. 데이터셋 다운로드
# 아래에 해당하는 내용을 자신의 데이터셋과 api key값에 맞게 수정할것 !!
import os
rf = Roboflow(api_key=os.environ["ROBOFLOW_API_KEY"])  # 키는 코드에 적지 말고 환경변수로 넘길 것
project = rf.workspace("minjun-kim-1s5r4").project("yolo-training-lii7y")
version = project.version(1)
dataset = version.download("yolov11")

# 3. YOLOv11 Nano 사전학습 모델 로드 및 파인튜닝 시작
model = YOLO("yolo11n.pt")  # COCO 기반 사전 가중치 자동 다운로드

results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=50,
    imgsz=640,
    batch=16
)
```
